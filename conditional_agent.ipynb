*** handling multiple nodes with routing node which will handle the flow ***

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END
from IPython.display import Image, display

class AgentState(TypedDict):
    number1: int
    number2: int
    operation1: str
    number3: int
    number4: int
    operation2: str
    finalNumber1: int
    finalNumber2: int

def addNode(state: AgentState) -> AgentState:
    state["result"] = state["item1"] + state["item2"]
    return state

def multiply(state: AgentState) -> AgentState:
    state["result"] = state["item1"] * state["item2"]
    return state

def condition_next_node(state: AgentState) -> AgentState:

    if(state["operation1"] == "add"):
        return "add_operation"

    elif(state["operation1"] == "multiply"):
        return "multiply_operation"

state_graph = StateGraph(AgentState)
state_graph.add_node("add_operation", addNode)
state_graph.add_node("multiply_operation", multiply)
state_graph.add_node("router", lambda state:state) # pass-through node to route to the next operation

```
